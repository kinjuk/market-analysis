# P/E mean reversion: standard test
Long when P/E **crosses below** its 5-year average, exit at **1.2x** that average, cash in between.
Compared with buy-and-hold of the same stock and with SPY or QQQ.

**Standard parameters:** 5y average · entry: cross below 1.0x · exit: 1.2x · 5 bps per side · filled one close after the signal · long-only, cash earns 0%.

Run from the `pe-strategies/` folder. Backtests use **trailing P/E** (point-in-time EPS from SEC EDGAR, see `trailing_pe.py`); the live signal uses forward P/E, which is still accumulating. See `strategy.md` for the caveats.

```python
import sys
sys.path.insert(0, ".")
%matplotlib inline
from report import run, plot, summary, trades

TICKER, BENCHMARK = "AAPL", "SPY"   # BENCHMARK: "SPY" or "QQQ"
SOURCE = "trailing"                 # "trailing" (builds data/trailing_<TICKER>.csv, needs SEC_USER_AGENT), "forward", or "csv"
```

```python
sig, bt, rets, P = run(TICKER, BENCHMARK, SOURCE)
fig = plot(sig, bt, rets, P, TICKER, BENCHMARK)
```

```python
summary(bt, rets).round(3)
```

```python
trades(bt).round(3)
```

**Read this before drawing conclusions:** fewer than ~5 trades is an anecdote, not evidence. A single ticker chosen today carries survivorship bias, and the trailing-P/E result does not validate the forward-P/E signal.
